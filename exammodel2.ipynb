```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
data={
    'ename':['rohan','amar','ashwini','lohith','mohan','pramod'],
    'department':['CS','CS','EC','EC','CS','EC'],
    'experience':[10,20,15,14,9,8],
    'salary':[450000,50000,30000,25000,10000,40000]
}
df=pd.DataFrame(data)
print(df)
table=pd.pivot_table(df,index=['department'],
values='salary',aggfunc=['sum'])
print(table)
table1=pd.pivot_table(df,index=['department'],
values='salary',aggfunc=['mean'])
print(table1)
```
