```python
import pandas as pd
df=pd.read_csv('/content/HSI.csv')
print(df)
print(df.head())
print(df.shape)
print(df.columns)
print(df.info())
print(df.describe())
print(df.isnull().sum())
print(df.select_dtypes(include='object').columns)
print(df['High'].unique())
num_cols = df.select_dtypes(include='number').columns
print(num_cols)
df[num_cols] = df[num_cols].fillna(df[num_cols].median())
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df['High_Encoded']=le.fit_transform(df['High'])
df=pd.get_dummies(df,columns=['High'],dtype=int)
print(df)
x=df.drop('Close',axis=1)
print(x)
y=df['Close']
print(y)
```
