```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
data={
    'Number':[1,2,3,4,5],
    'Pencil':[300,350,400,500,520],
    'Textbook':[250,350,400,420,500],
    'Drawing sheet':[100,200,200,250,300],
    'Totalunits':[800,1000,1320,1510,2000],
    'Profit':[8000,9500,10000,12000,18000]}
df=pd.DataFrame(data)
print(df)
df.info()
plt.plot(df['Number'],df['Profit'])
plt.xlabel('Number')
plt.ylabel('TotalProfit')
print("missing values")
print(df.isnull().sum())
print('sum of TotalProfit')
print(df['Profit'].sum())
print('max value of Drawing sheet column')
print(df['Drawing sheet'].max())
print('min value of Textbook column')
print(df['Textbook'].min())
new_Totalunits=df['Totalunits']+5
print(new_Totalunits)
```
